# แบบทดสอบเก็บคะแนน ครั้งที่ 2

**10 คะแนน**

## ให้สร้าง Web Application ระบบ RAG (Retrieval-Augmented Generation)


**คำชี้แจง**
ให้นักศึกษาพัฒนา Web Application แชตบอตที่ตอบคำถามจากคลังเอกสารความรู้ (ไทย/อังกฤษ) ด้วยเทคนิค RAG โดยให้นักศึกษาคิดหัวข้อที่สนใจได้เอง

นักศึกษาสามารถใช้ AI เป็นผู้ช่วยในการเขียนโปรแกรมในครั้งนี้ได้

**เงื่อนไข**

นักศึกษาจะต้อง นำโค้ดทั้งหมดขึ้น GitHub Repository ส่วนตัว และทำการ Deploy แอปพลิเคชันผ่าน Streamlit Community Cloud เพื่อให้สามารถเข้าใช้งานและแสดงผลผ่าน URL หน้าเว็บจริงได้

**ห้ามอัปโหลด API Key ขึ้น GitHub โดยเด็ดขาด** ให้เก็บ API Key ไว้ใน Secrets ของ Streamlit Community Cloud ตัวอย่างถ้าใช้ groq ก็จะเรียกใช้ด้วย `st.secrets["GROQ_API_KEY"]` (งานที่พบ API Key ใน Repository จะถูกหักคะแนน)

**ตัวอย่างแนวทางหัวข้อ**
1. ผู้ช่วยกฎหมายสำหรับประชาชน --> ตอบคำถามเรื่อง PDPA หรือกฎหมายแรงงานจากตัวบทกฎหมาย
2. ไกด์ท่องเที่ยวประจำจังหวัด --> แนะนำสถานที่เที่ยว ของดี เวลาเปิด-ปิด จากข้อมูลท่องเที่ยวของจังหวัด
3. ผู้ช่วยตอบคำถามคู่มือสินค้า --> ตอบวิธีใช้งาน การแก้ปัญหาเบื้องต้น และเงื่อนไขการรับประกันจากคู่มือสินค้า

*NOTE: นักศึกษาสามารถใช้ AI ในการสร้างเนื้อหาได้*


**ข้อกำหนดทางเทคนิค**

ให้นำเทคนิคจากที่เรียนมา มาประยุกต์ใช้ให้ครบทุกข้อ ได้แก่

- Document Loading & Chunking --> โหลดเอกสาร ทำความสะอาดข้อความ และแบ่งเอกสารเป็น Chunk ที่เหมาะสม

- Embedding & Vector Search --> แปลง Chunk เป็นเวกเตอร์ด้วย Sentence Embedding และค้นหาด้วย Vector Database (เช่น FAISS)

- Prompt Engineering --> ออกแบบ Prompt ให้ LLM ตอบจาก Context ที่ค้นได้เท่านั้น อ้างอิงแหล่งที่มา และตอบว่า "ไม่พบข้อมูล" เมื่อเอกสารไม่มีคำตอบ

- Large Language Model --> เรียกใช้ LLM ผ่าน API (เช่น Groq API) เพื่อสร้างคำตอบ

- Chatbot Interface --> สร้างหน้าแชตที่คุยต่อเนื่องได้ และแสดงเอกสารอ้างอิงที่ใช้ตอบทุกครั้ง

**สิ่งที่ต้องส่ง**

1. ส่งไฟล์ colab นี้กลับคืนมาโดยระบุ
    * หัวข้อที่เลือกมาใช้
    * URL หน้าเว็บ Streamlit --> ลิงก์แอปพลิเคชันที่ Deploy บน Streamlit Community Cloud ที่เปิดใช้งานได้จริง

    * GitHub Repository URL --> ลิงก์ซอร์สโค้ดระบบ ซึ่งต้องมีโครงสร้างไฟล์ครบถ้วน ดังนี้:

        - app.py (ไฟล์หลักของ Streamlit App)

        - requirements.txt (ระบุไลบรารีที่ต้องใช้ เช่น streamlit, sentence-transformers, faiss-cpu, groq, pythainlp)

        - README.md (อธิบายวิธีใช้งาน, แนวคิดของ Domain, แหล่งที่มาของเอกสาร และตัวอย่าง Prompt ที่ใช้สั่ง AI)

        - โฟลเดอร์ `data/` เก็บเอกสารความรู้ อย่างน้อย 10 ไฟล์ หรือรวมไม่น้อยกว่า 15,000 ตัวอักษร

        - ไฟล์ `test_questions.csv` คำถามทดสอบอย่างน้อย 10 ข้อ พร้อมคำตอบที่ถูกต้อง โดยต้องมีคำถามที่ไม่มีคำตอบในเอกสารอย่างน้อย 2 ข้อ

2. ไฟล์ PDF ที่รวบรวมการ Capture หน้าจอการทำงานของหน้าเว็บ พร้อมคำอธิบาย

**ข้อควรระวังในการ Deploy**

- Streamlit Community Cloud มีหน่วยความจำจำกัด ควรใช้ Embedding Model ขนาดเล็ก และโหลดโมเดลและ Index เพียงครั้งเดียว

- เพิ่มไฟล์ `.streamlit/secrets.toml` ไว้ใน `.gitignore` เพื่อไม่ให้ API Key หลุดขึ้น GitHub

- ทดสอบให้แน่ใจว่าเปิด URL จากเครื่องอื่นหรือโหมดไม่ระบุตัวตน (Incognito) แล้วใช้งานได้จริง

**เกณฑ์การให้คะแนน**

1. ไฟล์เอกสารความรู้และไฟล์คำถามทดสอบ (1 คะแนน)

2. นำโค้ดขึ้น GitHub และหน้าเว็บ Streamlit สามารถเข้าใช้งานได้จริง (2 คะแนน)

3. ความคิดสร้างสรรค์ในการเลือกหัวข้อ และความเหมาะสมของเอกสารความรู้ที่ใช้ (2 คะแนน)

4. การนำเทคนิค RAG มาประยุกต์ใช้ได้อย่างถูกต้อง ทั้ง Chunking, Retrieval และ Prompt (3 คะแนน)

5. ตอบคำถามโดยอิงจากเอกสาร แสดงแหล่งอ้างอิง และปฏิเสธได้เมื่อไม่มีข้อมูล (1 คะแนน)

6. หน้าเว็บใช้งานง่าย (1 คะแนน)
